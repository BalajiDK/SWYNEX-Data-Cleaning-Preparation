### Persisting Data using Google Drive
By mounting Google Drive, you can save your dataset permanently so it remains available even after the runtime terminates.

```python
from google.colab import drive
import shutil
import os

# Mount Google Drive
drive.mount('/content/drive')

# Create a folder in your Drive for datasets if it doesn't exist
destination_dir = '/content/drive/MyDrive/Colab_Datasets'
os.makedirs(destination_dir, exist_ok=True)

# Copy the dataset from local runtime to Google Drive
source_path = '/content/marketing_campaign.csv'
dest_path = os.path.join(destination_dir, 'marketing_campaign.csv')

if os.path.exists(source_path):
    shutil.copy(source_path, dest_path)
    print(f'File successfully saved to Google Drive at: {dest_path}')
else:
    print('Source file not found. Please make sure marketing_campaign.csv is uploaded.')
```

```python

df = pd.read_csv("marketing_campaign.csv", sep="\t")
df.head()
```

```python
print("dataset shape:",df.shape)
```

```python
print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])
```

```python
df.info()
```

```python
print("missing values in each colmn")
print(df.isnull().sum())
```

```python


missing_income = df[df["Income"].isnull()]

print("Number of rows with missing Income:",
      len(missing_income))

display(missing_income.head(10))
```

```python

print(df["Income"].describe())
print("Median income:", df["Income"].median())
```

```python


missing_before = df["Income"].isnull().sum()
median_income = df["Income"].median()
df["Income"] = df["Income"].fillna(median_income)
print("Missing values before cleaning:", missing_before)
print("Median income used:", median_income)
print("Missing values after cleaning:", df["Income"].isnull().sum())
```

```python
print("total missing values:",df.isnull().sum())
print("dataset shape:",df.shape)
```

```python

print("Duplicate rows before removal:", df.duplicated().sum())

df = df.drop_duplicates()
print("Duplicate rows after removal:", df.duplicated().sum())
print("Dataset shape after duplicate removal:", df.shape)
```

```python
print("Before:", df["Dt_Customer"].dtype)
df["Dt_Customer"] = pd.to_datetime(
    df["Dt_Customer"],
    dayfirst=True,
    errors="coerce"
)

print("After:", df["Dt_Customer"].dtype)
print("Invalid dates:", df["Dt_Customer"].isnull().sum())
```

```python

print("Education categories:")
print(df["Education"].value_counts())

print("\nMarital status categories:")
print(df["Marital_Status"].value_counts())
```

```python

unusual_status = ["Alone", "Absurd", "YOLO"]

display(
    df[df["Marital_Status"].isin(unusual_status)]
    [["ID", "Year_Birth", "Education", "Marital_Status", "Income"]]
)
```

```python

df["Marital_Status"] = df["Marital_Status"].replace({
    "Absurd": "Unknown",
    "YOLO": "Unknown"
})
print(df["Marital_Status"].value_counts())

print("\nRemaining unusual values:")
print(
    df[df["Marital_Status"].isin(["Absurd", "YOLO"])]
    ["Marital_Status"].count()
)
```

```python

print("Final dataset shape:", df.shape)
print("Total missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Invalid dates:", df["Dt_Customer"].isnull().sum())
df.to_csv("marketing_campaign_cleaned.csv", index=False)
print("\nCleaned dataset saved successfully!")
```
